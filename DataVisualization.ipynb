```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Read a sample S&P500 da
df=pd.read_csv("https://datahub.io/core/s-and-p-500/r/data.csv")
```

```python
#Filter the rows to only include records since 2011

#First convert the Date column to date & time format
df['Date'] = pd.to_datetime(df['Date'])  

#specify the start date: yyyy-mm-dd
start_date='2011-01-01'

#select rows that meet the criteria
mask = (df['Date'] >= start_date)

#create a new data frame which meets the selection criteria
df_new=df.loc[mask]
```

```python
#add a new column to categorize the earnings as high or low
def conditions(s):
    if (s['Earnings'] >  84):
        return "High"
    else:
        return "Low"

df_new['Class'] = df_new.apply(conditions, axis=1)
```

```python
# plot boxplot with x-axis as Class and y=Earnings
sns.boxplot(y=df_new["Earnings"])
```

```python
# plot boxplot with x-axis as Class and y=Earnings
sns.boxplot(y=df_new["Earnings"],x=df_new["Class"])
```

```python
#change colors 

sns.boxplot(y=df_new["Earnings"],x=df_new["Class"],palette="Blues")
```

```python
# scatter plot using Seaborn
ax = sns.scatterplot(x="Earnings", y="SP500", data=df_new)
plt.show()
```

```python
# scatter plot using Seaborn
ax = sns.lmplot(x="Earnings", y="SP500", data=df_new)
plt.show()
```

```python
#scatter plot using Matplotlib

fig, ax = plt.subplots(figsize=(5, 2.7), layout='constrained')
ax.scatter('Earnings', 'SP500',  data=df_new)
ax.set_xlabel('SP500')
ax.set_ylabel('Earnings')
```
