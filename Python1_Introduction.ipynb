```python
##String: ""
var_string="My name is John Fleming!"
print(var_string)
```

```python
#List: []

var_list=["John","Fleming"]  #position 0: John; position 1: Fleming
print("My name is "+var_list[0]+" "+var_list[1])
```

```python
#Join the elements of a list
print("My name is "+" ".join((var_list)))
```

```python
#Add new elements to a list
var_list.append("Jr.")
print("My name is "+" ".join((var_list)))
```

```python
#Remove an element from a list using the del function
del var_list[1] # or use var_list.pop(1)
print("My name is "+" ".join((var_list)))
```

```python
var_list[1]="Smith"
print("My name is "+var_list[0]+" "+var_list[1])
```

```python
#Tuples: ()

var_tuples=("John", "Fleming")

print("My name is "+var_tuples[0]+" "+var_tuples[1])
```

```python
var_tuples[1]="Smith"
print("My name is "+var_tuples[0]+" "+var_tuples[1])
```

```python
#Dictionary: key:value pairs

favorite_sports = {'Ralph Williams' : 'Football',
                   'Michael Tippett' : 'Basketball',
                   'Edward Elgar' : 'Baseball',
                   'Rebecca Clarke' : 'Netball',
                   'Ethel Smyth' : 'Badminton',
                   'Frank Bridge' : 'Rugby'}

```

```python
#Print the favorit sport for Rebecca Clarke
print(favorite_sports['Rebecca Clarke'])
```

```python
#Print the favorit sport for Frank Bridge
print(favorite_sports['Frank Bridge'])
```

```python
#Update value for Frank Bridge
favorite_sports['Frank Bridge']="Football"
```

```python
#Print the new key:value pair
print(favorite_sports['Frank Bridge'])
```
