```python
#Filter the rows to only include records since 2011

#First convert the Date column to date & time format
df['Date'] = pd.to_datetime(df['Date'])  

#specify the start date: yyyy-mm-dd
start_date='2011-01-01'

#select rows that meet the criteria
mask = (df['Date'] >= start_date)

#create a new data frame which meets the selection criteria
df_new=df.loc[mask]

#Print the top 5 rows in the new data table
df_new.head()
```

```python
#get the mean of SP500 using loc
round(df_new.loc[:,"SP500"].mean(),2)
```

```python
#get the mean of SP500 using iloc
round(df_new.iloc[:,1].mean(),2)
```

```python
#get the mean of SP500 using the column name
round(df_new["SP500"].mean(),2)
```

```python
#get the mean of SP500 when earnings are greater than 80
round(df_new[df_new['Earnings'] >= 80]["SP500"].mean(),2)
```

```python
#add a new column to categorize the earnings as high or low
def conditions(s):
    if (s['Earnings'] >  84):
        return "High"
    else:
        return "Low"

df_new['Class'] = df_new.apply(conditions, axis=1)
```

```python
#Dislay the new data frame
df_new.head()
```

```python
#Generate a pivot table summarizing SP500 values by High or Low class
import numpy as np

df_new.pivot_table(index='Class',values='SP500',aggfunc=np.mean)
```
